```python
import os
import gc
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.ticker as ticker
```

```python
#2.5 Top 10 hot brands total distribution analysis

input_path = r"C:\Users\punisher\Desktop\ecommerce_oct\top10_brands\csv\part-00000-84b77c0a-4f6b-4052-b722-a34ac75cf682-c000.csv"
hot_brand = pd.read_csv(input_path)

plt.figure(figsize=(14, 8), facecolor="lightgray")
ax = plt.gca()
ax.set_facecolor("white")


colors = list(plt.cm.Set3(np.linspace(0, 1, len(hot_brand))))

# barchart plot
bars = sns.barplot(x='brand', 
                   y='count', 
                   hue='brand',  
                   data=hot_brand, 
                   palette=colors,
                   ax=ax,
                   edgecolor='black',
                   linewidth=0.5,
                   alpha=0.9,
                   legend=False)  

# label
for container in bars.containers:
    ax.bar_label(container, 
                 fmt='{:,.0f}',      
                 label_type='edge', 
                 padding=5,
                 fontsize=13)
    
#thousands separator set
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x:,.0f}'))

#title beautification
plt.title('Top 10 Most Popular Brands', 
          fontsize=25, 
          fontweight='bold', 
          pad=20)

plt.xlabel('Brand', 
           fontsize=13, 
           labelpad=20)

plt.ylabel('Number of Products', 
           fontsize=13, 
           labelpad=20)

# gridline
ax.grid(True, 
        axis='y',
        alpha=0.3,  
        linewidth=0.5)

# remove  boundline
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)


ax.spines['left'].set_color('black')
ax.spines['bottom'].set_color('black')
ax.spines['left'].set_linewidth(1)
ax.spines['bottom'].set_linewidth(1)


plt.tight_layout()
plt.subplots_adjust(bottom=0.1)  
plt.savefig(' Bar Chart of Top 10 Most Popular Brands.png')
plt.show()
plt.close()
```

```python
input_path = r"C:\Users\punisher\Desktop\ecommerce_oct\top10_brands\csv\part-00000-84b77c0a-4f6b-4052-b722-a34ac75cf682-c000.csv"
hot_brand = pd.read_csv(input_path)

plt.figure(figsize=(14, 10), facecolor="lightgray")
ax = plt.gca()
ax.set_facecolor("lightgray")


colors = list(plt.cm.Set3(np.linspace(0, 1, len(hot_brand))))

# percentage calculate
total = hot_brand['count'].sum()
percentages = (hot_brand['count'] / total * 100).round(1)

# explode biggest parts 
explode = [0.1 if i == 0 else 0.05 if i < 3 else 0 for i in range(len(hot_brand))]

# piechart plot
wedges, texts, autotexts = ax.pie(hot_brand['count'],
                                  labels=hot_brand['brand'],
                                  autopct=lambda pct: f'{pct:.1f}%',
                                  colors=colors,
                                  explode=explode,
                                  startangle=90,
                                  shadow=True,  # shades
                                  wedgeprops={'edgecolor': 'black', 
                                             'linewidth': 1.5,
                                             'alpha': 0.85},
                                  textprops={'fontsize': 11, 
                                            'fontweight': 'bold'},
                                  pctdistance=0.8)


# beautify percentage number
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontweight('bold')
    autotext.set_fontsize(10)

# beautify label
for text in texts:
    text.set_fontsize(11)

plt.title('Top 10 Most Popular Brands Distribution\n', 
          fontsize=16, 
          fontweight='bold', 
          pad=20,)

# add center circle
centre_circle = plt.Circle((0,0), 0.70, fc='white', edgecolor='gray')
ax.add_artist(centre_circle)

# total count in the center
ax.text(0, 0, f'Total\n{total:,}', 
        ha='center', va='center', 
        fontsize=12, fontweight='bold',
        color='crimson')


plt.tight_layout()
plt.savefig('Pie Chart of Top 10 Most Popular Brands.png')
plt.show()
plt.close()
```

```python
#2.6 Weekly Sales Trends Analysis

input_path = r"C:\Users\punisher\Desktop\ecommerce_oct\weekly_sales\csv\part-00000-e366f07f-dd25-4160-9d90-0eeac953a613-c000.csv"
total_sales = pd.read_csv(input_path)

plt.figure(figsize=(14, 8), facecolor="lightgray")
ax = plt.gca()
ax.set_facecolor("lightgray")

# Line chart
line = sns.lineplot(x=total_sales.index+1,
                   y=total_sales.sales, 
                   marker='o',           # bullets 
                   markersize=8,
                   linewidth=2.5,
                   color='#2E86AB',     
                   markerfacecolor='white',
                   markeredgecolor='#2E86AB',
                   markeredgewidth=2)

# Add data labels
for x, y in zip(total_sales.index+1, total_sales.sales):
    plt.text(x, y + y * 0.02,           # Slightly move upwards
             f"{int(round(y)):,}",       # Thousand Separator Formatting
             ha='center', 
             va='bottom', 
             fontsize=11,
             fontweight='bold',
             color='darkblue',
             bbox=dict(boxstyle="round,pad=0.3", 
                      facecolor='lightyellow', 
                      edgecolor='gray', 
                      alpha=0.7))

# Set title & tags
plt.title('Weekly Sales Trends Within Months', 
          fontsize=16, 
          fontweight='bold', 
          pad=20,
          color='darkred')

plt.ylabel('Total Sales', 
          fontsize=13, 
          fontweight='bold',
          labelpad=15)

plt.xlabel('Week-in-Month', 
          fontsize=13, 
          fontweight='bold',
          labelpad=15)

# Set x-axis ticks
plt.xticks(ticks=range(1, 6), fontsize=11)
plt.yticks(fontsize=11)

# Set the y-axis to use thousand separators
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x:,.0f}'))

#  gridlines
ax.grid(True, 
        axis='both', 
        alpha=0.3, 
        linestyle='--', 
        linewidth=0.5)

# Remove the top and right border lines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Set the border color and thickness
ax.spines['left'].set_color('gray')
ax.spines['bottom'].set_color('gray')
ax.spines['left'].set_linewidth(0.5)
ax.spines['bottom'].set_linewidth(0.5)


# padding
ax.fill_between(total_sales.index+1, 
                total_sales.sales, 
                alpha=0.2, 
                color='#2E86AB')

plt.tight_layout()
plt.subplots_adjust(bottom=0.1)
plt.savefig('Line chart of Weekly Sales Trends Within Months.png')
plt.show()
plt.close()
```

```python
# 2.6.1 Weekly Sales Trend Analysis of the Top 10 Popular Product Categories
input_path = r"C:\Users\punisher\Desktop\ecommerce_oct\weekly_sales_top_categories\csv\part-00000-ea949a4e-8b71-429e-a79d-734b8713f5ee-c000.csv"
product_sales = pd.read_csv(input_path)

plt.figure(figsize=(16, 10), facecolor="lightgray")
ax = plt.gca()
ax.set_facecolor("lightgray")

# Use Set3 color scheme for consistency
colors = list(plt.cm.Set3(np.linspace(0, 1, len(product_sales['category_code'].unique()))))

# Enhanced line plot with consistent styling
sns.lineplot(data=product_sales, 
             x='week_in_month', 
             y='sales', 
             hue='category_code', 
             marker='o',           # Circular markers
             markersize=6,
             linewidth=2,
             palette=colors)       # Consistent color palette

# Add data labels with improved styling
for category_code, group in product_sales.groupby('category_code'):
    # Peak value label
    max_idx = group['sales'].idxmax()
    x = group.loc[max_idx, 'week_in_month']
    y = group.loc[max_idx, 'sales']
    plt.text(x, y + y * 0.03,           # Slight upward offset
             f"{int(round(y)):,}",       # Thousand separator formatting
             ha='center', 
             va='bottom', 
             fontsize=9,
             fontweight='bold',
             color='darkblue',
             bbox=dict(boxstyle="round,pad=0.2", 
                      facecolor='lightyellow', 
                      edgecolor='gray', 
                      alpha=0.7))

    # Lowest value label
    min_idx = group['sales'].idxmin()
    x = group.loc[min_idx, 'week_in_month']
    y = group.loc[min_idx, 'sales']
    plt.text(x, y - y * 0.03,           # Slight downward offset
             f"{int(round(y)):,}", 
             ha='center', 
             va='top', 
             fontsize=9,
             fontweight='bold',
             color='darkred',
             bbox=dict(boxstyle="round,pad=0.2", 
                      facecolor='lightcoral', 
                      edgecolor='gray', 
                      alpha=0.7))

# Enhanced title and axis labels
plt.title('Weekly Sales Trends Within the Month (Top 10 Product Categories)', 
          fontsize=16, 
          fontweight='bold', 
          pad=20,
          color='darkred')

plt.xlabel('Week in Month', 
          fontsize=13, 
          fontweight='bold',
          labelpad=15)

plt.ylabel('Total Sales', 
          fontsize=13, 
          fontweight='bold',
          labelpad=15)

# Format axes
plt.xticks(ticks=range(1, 6), fontsize=11)
plt.yticks(fontsize=11)

# Thousand separator formatting for y-axis
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x:,.0f}'))

# Enhanced grid styling
ax.grid(True, 
        axis='both', 
        alpha=0.3, 
        linestyle='--', 
        linewidth=0.5)

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Style remaining spines
ax.spines['left'].set_color('gray')
ax.spines['bottom'].set_color('gray')
ax.spines['left'].set_linewidth(0.5)
ax.spines['bottom'].set_linewidth(0.5)

# Enhanced legend positioning
plt.legend(bbox_to_anchor=(1.05, 1), 
           loc='upper left', 
           title='Product Categories',
           title_fontsize=12,
           fontsize=10,
           frameon=True,
           fancybox=True,
           shadow=True,
           framealpha=0.9)

plt.tight_layout()
plt.subplots_adjust(right=0.85)  # Adjust for legend space
plt.savefig('Line chart of Weekly Sales Trends Within the Month (Top 10 Product Categories).png')
plt.show()
plt.close()
```

```python
# 2.8 Daily Sales Trends Analysis
input_path = r"C:\Users\punisher\Desktop\ecommerce_oct\daily_sales\csv\part-00000-25aefec8-6790-43a2-89bd-a50d274d2ef7-c000.csv"
total_sales = pd.read_csv(input_path)

plt.figure(figsize=(16, 8), facecolor="lightgray")
ax = plt.gca()
ax.set_facecolor("lightgray")

# Enhanced line plot with consistent styling
line = sns.lineplot(x=total_sales.index, 
                   y=total_sales.sales,
                   linewidth=2.5,
                   color='#2E86AB')      # Consistent blue color

# Find peak and valley values
max_idx = total_sales['sales'].idxmax()
min_idx = total_sales['sales'].idxmin()
max_val = total_sales.loc[max_idx, 'sales']
min_val = total_sales.loc[min_idx, 'sales']

# Highlight extreme points with enhanced markers
plt.scatter(max_idx, max_val, color='red', s=100, zorder=5, 
           label=f'Peak: Day {max_idx}', marker='D', edgecolor='darkred', linewidth=2)
plt.scatter(min_idx, min_val, color='blue', s=100, zorder=5, 
           label=f'Valley: Day {min_idx}', marker='s', edgecolor='darkblue', linewidth=2)

# Enhanced annotations for extreme values
plt.text(max_idx, max_val + max_val * 0.03, 
         f"Peak: {int(round(max_val)):,}", 
         ha='center', va='bottom', 
         fontsize=11, fontweight='bold', color='darkred',
         bbox=dict(boxstyle="round,pad=0.3", facecolor='lightcoral', 
                  edgecolor='red', alpha=0.8))

plt.text(min_idx, min_val - min_val * 0.03, 
         f"Valley: {int(round(min_val)):,}", 
         ha='center', va='top', 
         fontsize=11, fontweight='bold', color='darkblue',
         bbox=dict(boxstyle="round,pad=0.3", facecolor='lightblue', 
                  edgecolor='blue', alpha=0.8))

# Add fill under the line for better visualization
ax.fill_between(total_sales.index, 
                total_sales.sales, 
                alpha=0.2, 
                color='#2E86AB')

# Enhanced title and labels
plt.title('Daily Sales Trends for the current month', 
          fontsize=16, fontweight='bold', pad=20, color='darkred')

plt.ylabel('Total Sales', 
          fontsize=13, fontweight='bold', labelpad=15)

plt.xlabel('Day of Month', 
          fontsize=13, fontweight='bold', labelpad=15)

# Format x-axis
plt.xticks(ticks=total_sales.index[::2],  # Show every other day to reduce clutter
          rotation=45, 
          fontsize=11)

# Thousand separator formatting for y-axis
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x:,.0f}'))

# Enhanced grid styling
ax.grid(True, axis='both', alpha=0.3, linestyle='--', linewidth=0.5)

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Style remaining spines
ax.spines['left'].set_color('gray')
ax.spines['bottom'].set_color('gray')
ax.spines['left'].set_linewidth(0.5)
ax.spines['bottom'].set_linewidth(0.5)

# Enhanced legend
plt.legend(loc='upper right', 
          fontsize=11,
          frameon=True,
          fancybox=True,
          shadow=True,
          framealpha=0.9)

# Add statistical information
mean_sales = total_sales['sales'].mean()
plt.axhline(y=mean_sales, color='orange', linestyle='--', alpha=0.7, 
           label=f'Average: {int(round(mean_sales)):,}')

# Update legend to include average line
plt.legend(loc='upper right', 
          fontsize=11,
          frameon=True,
          fancybox=True,
          shadow=True,
          framealpha=0.9)


plt.tight_layout()
plt.subplots_adjust(bottom=0.15)  # Adjust for rotated labels
plt.show()

# Clean up memory
del max_idx, min_idx, max_val, min_val, total_sales, ax, mean_sales
plt.savefig('Scatter Plot of Daily Sales Trends for the current month.png')
plt.close()
gc.collect()
```
